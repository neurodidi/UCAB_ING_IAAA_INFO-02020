# Numpy: arrays of data

This data structures are lists of data that may ressemble good ol' vectors and matrices (and higher dimensional tensor).

They are fundamental for data analysis and ML. On one hand, they facilitate mathematical operations along lists of data. On the other hand, it is the basic element for data containing in fnctions from libraries like scikit-learn.

````python
random.seed(42) # This we use to always get the same randomly generated numbers

x = random.sample(range(101), k=10)
print(f"Generated list: {x}")
print(type(x))
# Now we will get our list to become a numpy array
print()

x = np.array(x)
print(f"Numpy array: {x}")
print(type(x))

print()

# let us create a second np.array
y = [np.random.randint(1, 100) for _ in range(10)] # just another way to create a list of random integers
y = np.array(y)
print(f"Second array: {y}")
print(type(y))
````

````python
# some operations with arrays:

print(x+y) # addition
print()
print(x-y) # substraction
print()
print(x*y) # multiplication
print()
print(x/y) # division
print()
print(x**y) # Powering

"""
Please note that these operations are element wise
"""
````

````python
# We can do the same with scalars

print(x+2) # addition
print()
print(x*2) # multiplication
print()
````

````python
# if the array represents data, we might wanto to get some statistics out of it
variables = [x, y]

for var in variables:
  print(f"stats for {var}")
  print("average ", np.average(var))
  print("std ", np.std(var))
  print("median ", np.median(var))
  print("max ", np.max(var))
  print("min ", np.min(var))
  print("sum ", np.sum(var))
  print()
````

````python
# We can create matrices as well

A = np.ones((4, 4))
print(A)
print()

# we can follow the same logic as before for matrix operations
print(A + 2)
print()
print(A/2)
print()

# Now, let's do good ol' matrix multiplication

z = random.sample(range(101), k=3)
B = A[0:3, 0:3] + 3  # here we are generating a new matrix with the first 3
# columns and rows of the A matrix
print(z)
print(B)

m = B@z

print(m)
````

````python
# Let us generate a random matrix

R = np.random.randint(1, 101, size=(2, 3))
print("2x3 Random Integer Matrix:")
display(R)
print()
# We can transpose it by using the .T append
print("Transposed R matrix")
display(R.T)
````

we also have a sublibrary fully dadicated to linear algebra within numpy

````python
# mangitud del vector m
print(np.linalg.norm(m))
````

````python
# we might want to solve a linear equation (as we will see for the regression algorithm)
# We solve for x in the equation: Ax = b

# Coefficients
A_sys = np.random.randint(1, 10, size=(4, 4))

# Constants
b_sys = np.random.randint(1, 10, size=4)

# We solve for x
x_sol = np.linalg.solve(A_sys, b_sys)

print("Matrix A (4x4):")
print(A_sys)
print("\nVector b:")
print(b_sys)
print("\nSolution x:")
print(x_sol)

# Verification: A @ x should equal b
print("\nVerification (A @ x):")
print(np.dot(A_sys, x_sol))
````

Finally, let us see how we can stick together np.arrays

````python
# To concatenate, the arrays must be passed as a list or tuple: [A, B]
# Both A and A_sys are 4x4, so they can be joined
result = np.concatenate([A, A_sys], axis=0) #default axis is 0 (vertical)
display(result)
print()

result = np.concatenate([A, A_sys], axis=1) #axis 1 is horizontal
display(result)
````

# Pandas: Datframes for data

A way to arrange a lot of data in columns is to set uo a dataframe. This are useful structures that we use to help storing and visualizing data.

Pandas has some useful function through which we can arrange and "juice out" information off the data we store.

````python
random.seed(42)

X1 = np.random.randn(100) # Generate a array of random numbers extracted from a Normal Standard distribution
# print(X1)
# type(X1) # Check if it is a np.prray

# Let us generate more data from other distributions

rng = np.random.default_rng()

X2 = rng.random(100) # random number from a uniform distribution between 0 and 1

mu, sigma = 1, 0.5 # mean and standard deviation
X3 = rng.normal(mu, sigma, 100) # ramdom numberrs froma normal distribution with mean in 1 and sigma 0.5

X4 = rng.integers(low=0, high=1000, size=100) # random numbers between 0 and 1000

a = 4. # shape
samples = 100


X5 = rng.standard_gamma(3., 100)
````

Now, let us create a new, Y variable, dependent of all previous X1 to X5 variables, but adding some random noise to it.

𝑌 = 𝚇1÷(𝚇2 - 𝚇3) ⋅ 5 + 𝚇4𝚇5 + ϵ



````python
Y = X1/(X2 - X3) * 5 + X4 * X5 + rng.standard_normal(100)

````

We can now put all together into a pandas dataframe

````python
df = pd.DataFrame(
    {
     "X1": X1,
     "X2": X2,
     "X3": X3,
     "X4": X4,
     "X5": X5,
     "Y": Y,
    })
````

let us assume we have data missing from our data set.

We would like to assess it and decide what to do about it.

````python


# Replace alues greater than 900 with an NAN value
df.loc[df['X4'] > 900, 'X4'] = np.nan

#differet tha .iloc

# How many NaN values do we have?
df.isna().sum()
````

````python
df_nan_rows = df[df.isna().any(axis=1)]
display(df_nan_rows)
````

We now identified that 15 of 100 data points have NaN values for the X4 variable. We now have to decide what to do with this missing values. Among many options, we might want to:
1. Generate a prediction model that predicts the value of X4 based on something else (like the rest of the X variables).
2. Replace with the mean X4 value.
3. Drop the rows completely (we might want to check for the importance of this data points beforehand).

In this case, we will just drop them.


````python
df.dropna(inplace=True)
df.isna().sum() # We now have no NaN values
df.info() # But now we have 15 less data points
````
