### 1.4 Functions, Reusability & Type Hints
Writing modular functions with type hints and docstrings prevents bugs in data engineering and ML pipelines.

````python
# Cell 1.4: Reusable Functions Demo
from typing import List, Dict, Union

def min_max_normalize(data: List[float]) -> List[float]:
    """
    Normalizes a numerical list to the range [0.0, 1.0].
    
    Formula: x_norm = (x - min) / (max - min)
    """
    min_val = min(data)
    max_val = max(data)
    
    if max_val == min_val:
        return [0.0 for _ in data]
        
    return [(x - min_val) / (max_val - min_val) for x in data]

raw_scores = [40.0, 60.0, 80.0, 100.0]
normalized = min_max_normalize(raw_scores)
print(f"Raw scores       : {raw_scores}")
print(f"Normalized scores: {normalized}")
````

### 2.1 The Need for Speed: Python Loops vs. Vectorization
Why can't we just use Python lists for machine learning? Let's run a performance benchmark using `%timeit`.

````python
# Cell 2.1: Speed Benchmark - Python Lists vs. NumPy ndarray
import time

N = 1_000_000
py_list_a = list(range(N))
py_list_b = list(range(N))

np_arr_a = np.arange(N)
np_arr_b = np.arange(N)

# 1. Pure Python Loop
t0 = time.perf_counter()
py_result = [a + b for a, b in zip(py_list_a, py_list_b)]
t_python = time.perf_counter() - t0

# 2. NumPy Vectorized Addition
t0 = time.perf_counter()
np_result = np_arr_a + np_arr_b
t_numpy = time.perf_counter() - t0

speedup = t_python / t_numpy
print(f"⏱️ Pure Python Loop Time : {t_python * 1000:.2f} ms")
print(f"⚡ NumPy Vectorized Time : {t_numpy * 1000:.2f} ms")
print(f"🚀 NumPy is approximately {speedup:.1f}x faster!")
````

### 2.4 Broadcasting & Matrix Algebra (Normal Equation Preview)

#### What is Broadcasting?
Broadcasting allows NumPy to perform arithmetic operations on arrays of different shapes without copying data.
For example: centering data by subtracting the mean feature vector:
$$\tilde{X} = X - \boldsymbol{\mu}$$
where $X \in \mathbb{R}^{N \times D}$ and $\boldsymbol{\mu} \in \mathbb{R}^{D}$.

#### Matrix Multiplication & Linear Systems
- Element-wise multiplication: `A * B`
- Matrix product (dot product): `A @ B` or `np.dot(A, B)`
- Solving $A \mathbf{w} = \mathbf{b}$: `np.linalg.solve(A, b)`

In Ordinary Least Squares (OLS) Linear Regression, the optimal weights $\mathbf{w}^*$ are given analytically by the **Normal Equation**:
$$\mathbf{w}^* = (X^T X)^{-1} X^T \mathbf{y} \iff (X^T X) \mathbf{w}^* = X^T \mathbf{y}$$

````python
# Cell 2.4: Matrix Algebra & Linear Systems
# Let's solve a 2D linear system:
# 2*w0 + 1*w1 = 8
# 1*w0 + 3*w1 = 14
A = np.array([[2.0, 1.0],
              [1.0, 3.0]])
b = np.array([8.0, 14.0])

# Solve for w:
w = np.linalg.solve(A, b)
print(f"Solved weights w: {w}")
print(f"Verification A @ w: {A @ w} (should match b: {b})")

# Vector Euclidean Norm (L2 norm)
v = np.array([3.0, 4.0])
norm_v = np.linalg.norm(v)
print(f"L2 Norm of [3, 4] = {norm_v:.2f}")
````

### 3.3 Selection: `.loc[]` vs. `.iloc[]` (The Critical Indexing Trap)

A classic source of bugs in data pipelines is confusing **label-based** indexing with **position-based** indexing:
- **`.loc[row_label, col_label]`**: Accesses by **Index Label** (the name of the index or column).
- **`.iloc[row_pos, col_pos]`**: Accesses by **Integer Position** (always $0, 1, 2, \dots, N-1$).

Let's demonstrate what happens when rows are dropped and indices become non-contiguous!

---

### 5.1 The Standard Machine Learning Pipeline

```
┌───────────────┐      ┌─────────────────────────┐      ┌─────────────────────────┐
│ Raw DataFrame │ ───► │ Split Features & Target │ ───► │    Train / Test Split   │
│   (Pandas)    │      │    X (2D) and y (1D)    │      │ X_train, X_test, y_...  │
└───────────────┘      └─────────────────────────┘      └────────────┬────────────┘
                                                                     │
                                                                     ▼
┌───────────────┐      ┌─────────────────────────┐      ┌─────────────────────────┐
│  Evaluation   │ ◄─── │ Model Prediction        │ ◄─── │ Feature Scaling         │
│ (Acc, Report) │      │ y_pred = model.predict()│      │ scaler.fit_transform()  │
└───────────────┘      └─────────────────────────┘      └─────────────────────────┘
```

### 5.2 Feature Matrix ($X$) & Target Vector ($y$)
Scikit-Learn requires:
- **$X$ (Feature Matrix)**: A 2D array or DataFrame of shape `(n_samples, n_features)`.
- **$y$ (Target Vector)**: A 1D array or Series of shape `(n_samples,)` containing the ground truth labels.

````python
# Cell 5.1: Feature Extraction
# Prepare clean modeling data
model_df = df[['survived', 'pclass', 'age', 'sibsp', 'parch', 'fare']].copy()

# Impute any remaining missing values in age with median
model_df['age'] = model_df['age'].fillna(model_df['age'].median())

# Separate X and y
feature_names = ['pclass', 'age', 'sibsp', 'parch', 'fare']
X = model_df[feature_names]
y = model_df['survived']

print(f"Feature Matrix X shape : {X.shape}  (2D)")
print(f"Target Vector y shape  : {y.shape}   (1D)")
print("\nFirst 3 rows of X:")
display(X.head(3))
````

### 5.3 Train-Test Split & Data Leakage Prevention
Why do we split data? If we evaluate a model on the same data it was trained on, it can simply memorize the inputs (**overfitting**). We must test on **unseen test data** to estimate true generalization performance.

> [!IMPORTANT]
> **Data Leakage Rule:** Any preprocessing transformation (such as mean/std scaling) must be computed **ONLY on `X_train`**, and then applied to `X_test`. Never fit scalers on the entire dataset!

````python
# Cell 5.2: Train/Test Split & Feature Scaling
# Split: 80% Training, 20% Testing (stratified by class distribution)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y
)

print(f"Training samples : {X_train.shape[0]}")
print(f"Testing samples  : {X_test.shape[0]}")

# Feature Standardization: z = (x - mu) / sigma
scaler = StandardScaler()

# Notice: fit_transform on TRAIN, but only transform on TEST!
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\nFeature scaling complete. Mean of train feature 0:", round(X_train_scaled[:, 0].mean(), 4))
print("Standard deviation of train feature 0        :", round(X_train_scaled[:, 0].std(), 4))
````

### 5.4 Model Training & Evaluation
We instantiate a `LogisticRegression` classifier, train it via `.fit()`, and evaluate its predictions with `.predict()`.

````python
# Cell 5.3: Model Training and Prediction
# 1. Instantiate the estimator
clf = LogisticRegression(random_state=42)

# 2. Fit the model to the training set
clf.fit(X_train_scaled, y_train)

# 3. Predict on unseen test data
y_pred = clf.predict(X_test_scaled)

# 4. Compute Metrics
test_acc = accuracy_score(y_test, y_pred)
print(f"🎯 Test Set Accuracy: {test_acc * 100:.2f}%\n")

print("--- Detailed Classification Report ---")
print(classification_report(y_test, y_pred, target_names=["Did not survive (0)", "Survived (1)"]))
````

````python
# Cell 5.4: Confusion Matrix Visualization
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 4.5))
sns.heatmap(
    cm, 
    annot=True, 
    fmt='d', 
    cmap='Blues', 
    xticklabels=["Pred: Died", "Pred: Survived"],
    yticklabels=["True: Died", "True: Survived"]
)
plt.title(f"Confusion Matrix (Test Accuracy: {test_acc*100:.1f}%)", fontweight='bold')
plt.ylabel("Actual Label")
plt.xlabel("Predicted Label")
plt.show()
````
